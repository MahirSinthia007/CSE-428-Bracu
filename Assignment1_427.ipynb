Numpy



```python
import numpy as np
```

```python
#1
arr1=np.zeros(20,dtype=int)
arr2=np.zeros(20,dtype=int)
for i in range(len(arr1)):
    arr1[i]=np.random.randint(1,101)
    arr2[i]=np.random.randint(1,101)
print(arr1)
print(arr2)
```

```python
#2
arr1.resize(5,4)
print(arr1)
arr2.resize(5,4)
print(arr2)
```

```python
#3
scalar_multiplier=np.random.randint(1,100)
print(f"This is the generated scalar multiplier {scalar_multiplier}")
a=arr1*scalar_multiplier
print(f"multiplied matrix 1: {a}")
b=arr2*scalar_multiplier
print(f"multiplied matrix 2: {b}")
```

```python
#4
b=b.T
print(b)# transposed so that it can be multiplied
print(b.shape)

c=np.matmul(a,b)
print("Multiplied Matrix C")
print(f"{c}")
```

```python
#5
print("The columns 2 and 3: ")
print(c[:,2:4])
print("The whole matrix: ")
print(c[:,:])
```

```python
#6
max=float("-inf")
max_index_r=0
max_index_c=0
r_limit,c_limit=c.shape
print(r_limit,c_limit)

for row in range(0,r_limit):
    for col in range(0,c_limit):
        if c[row][col]>=max:
            max=c[row][col]
            max_index_r=row
            max_index_c=col
print(f"{max} is the largest element")
print(f"index in the matrix are: {max_index_r},{max_index_c}")


#if used np.max and np.argmax

print(np.max(c))
print(np.argmax(c))
row1,col1=np.unravel_index(np.argmax(c),c.shape)
print(row1,col1)
```

```python
#7
c_new=c.flatten()
print(c_new)
print(f"The memory location of c_new {id(c_new)}")
print(f"The memory location of c {id(c)}")
```

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
```

```python
#1
x=temperature_dhaka
y=humidity_dhaka
plt.title("temperature vs humidity")
plt.xlabel("values of temperature of dhaka")
plt.ylabel("values of humidity of dhaka")
plt.grid()
plt.scatter(x, y, color = "red", marker = "*")
plt.show()
```
